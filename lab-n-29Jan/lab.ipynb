```python
from sklearn.datasets import load_breast_cancer
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
import plotly.express as px
import pandas as pd
```

```python
from sklearn import tree
import matplotlib.pyplot as plt
```

```python
data2 = load_breast_cancer()
df = pd.DataFrame(data2.data, columns=data2.feature_names)
```

```python
df
```

```python
df['target'] = data2.target
df
```

```python
data = load_breast_cancer()
x = data.data
y = data.target
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
tree = DecisionTreeClassifier(random_state=0)
model = tree.fit(x_train, y_train)
```

```python
print(f'training accuracy: {tree.score(x_train, y_train)}')
print(f'testing accuracy: {tree.score(x_test, y_test)}')
```

```python
from sklearn.tree import export_graphviz
export_graphviz(tree, out_file="tree.dot",
                feature_names = data2.feature_names,
                class_names = ['malignant', 'benign'],
                rounded = True, proportion = False,
                precision = 2, filled = True)
```

```python
import graphviz
with open("tree.dot") as f:
    dot_graph = f.read()
graphviz.Source(dot_graph)
```

# Pre-pruning

```python
train_accuracy = []
test_accuracy = []
for depth in range(1, 21):
    tree = DecisionTreeClassifier(max_depth=depth, random_state=0)
    model = tree.fit(x_train, y_train)
    train_accuracy.append(tree.score(x_train, y_train))
    test_accuracy.append(tree.score(x_test, y_test))
```

```python
px.line(x=list(range(1, 21)), y=[train_accuracy, test_accuracy])
```

```python
data2 = load_breast_cancer()
df = pd.DataFrame(data2.data, columns=data2.feature_names)

import numpy as np
import matplotlib.pyplot as plt

# pick the most important feature from your tree
X = df[["mean concave points"]].values.ravel()
Y = data2.target   # target (0 = malignant, 1 = benign)

# Fit 3 curves of different complexity
p1 = np.polyfit(X, Y, 1)    # underfit
p2 = np.polyfit(X, Y, 3)    # good fit
p3 = np.polyfit(X, Y, 10)   # overfit

x_smooth = np.linspace(X.min(), X.max(), 300)
y1 = np.polyval(p1, x_smooth)
y2 = np.polyval(p2, x_smooth)
y3 = np.polyval(p3, x_smooth)

plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.scatter(X, Y)
plt.plot(x_smooth, y1)
plt.title("1. Underfitting")

plt.subplot(1,3,2)
plt.scatter(X, Y)
plt.plot(x_smooth, y2)
plt.title("2. Good Fit")

plt.subplot(1,3,3)
plt.scatter(X, Y)
plt.plot(x_smooth, y3)
plt.title("3. Overfitting")

plt.tight_layout()
plt.show()
```

```python
print("Train:", model.score(x_train, y_train))
print("Test :", model.score(x_test, y_test))
```

```python
from sklearn.metrics import ConfusionMatrixDisplay
import matplotlib.pyplot as plt

y_pred = model.predict(x_test)

ConfusionMatrixDisplay.from_predictions(y_test, y_pred)
plt.title("Confusion Matrix")
plt.show()
```

```python
from sklearn.metrics import roc_curve, auc

y_prob = model.predict_proba(x_test)[:, 1]

fpr, tpr, _ = roc_curve(y_test, y_prob)
roc_auc = auc(fpr, tpr)

plt.plot(fpr, tpr, label=f"AUC = {roc_auc:.3f}")
plt.plot([0, 1], [0, 1], linestyle="--")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.legend()
plt.show()
```

```python
print("Train accuracy:", model.score(x_train, y_train))
print("Test accuracy :", model.score(x_test, y_test))
```
